```python
%sql
IF EXISTS (SELECT * FROM sys.external_data_sources WHERE name = 'capstone_blob')
    DROP EXTERNAL DATA SOURCE capstone_blob;
IF EXISTS (SELECT * FROM sys.database_scoped_credentials WHERE name = 'blob_cred')
    DROP DATABASE SCOPED CREDENTIAL blob_cred;
GO

CREATE DATABASE SCOPED CREDENTIAL blob_cred
WITH IDENTITY = 'SHARED ACCESS SIGNATURE',
     SECRET = 'PASTE_SAS_TOKEN_HERE';
GO

CREATE EXTERNAL DATA SOURCE capstone_blob
WITH (
    TYPE = BLOB_STORAGE,
    LOCATION = 'https://capstone03.blob.core.windows.net/capstone3flight',
    CREDENTIAL = blob_cred
);
GO
```
